1. Write a "counter" class that can be incremented up to a specified maximum value, will print an error if an attempt is made to increment beyond that value, and allows reseting the counter. 

```python
class Counter:
    def __init__(self, max_value):
        self.max_value = max_value
        self.count = 0
    
    def increment(self):
        if self.count >= self.max_value:
            print("Error: Counter is at maximum value")
        else:
            self.count += 1
    
    def reset(self):
        self.count = 0
    
    def get_count(self):
        return self.count
```

```python
c = Counter(5)
print("Count:", c.get_count())

c.increment()
c.increment()
print("Count:", c.get_count())

c.increment()
c.increment()
c.increment()
print("Count:", c.get_count())

c.increment()

c.reset()
print("After reset:", c.get_count())
```

2. Copy and paste your solution to question 1 and modify it so that all the data held by the counter is private. Implement functions to check the value of the counter, check the maximum value, and check if the counter is at the maximum.

```python
class Counter:
    def __init__(self, max_value):
        self.__max_value = max_value
        self.__count = 0
    
    def increment(self):
        if self.__count >= self.__max_value:
            print("Error: Counter is at maximum value")
        else:
            self.__count += 1
    
    def reset(self):
        self.__count = 0
    
    def get_count(self):
        return self.__count
    
    def get_max_value(self):
        return self.__max_value
    
    def is_at_max(self):
        return self.__count >= self.__max_value
```

3. Implement a class to represent a rectangle, holding the length, width, and $x$ and $y$ coordinates of a corner of the object. Implement functions that compute the area and perimeter of the rectangle. Make all data members private and privide accessors to retrieve values of data members. 

```python
class Rectangle:
    def __init__(self, x, y, length, width):
        self.__x = x
        self.__y = y
        self.__length = length
        self.__width = width
    
    def area(self):
        return self.__length * self.__width
    
    def perimeter(self):
        return 2 * (self.__length + self.__width)
    
    def get_x(self):
        return self.__x
    
    def get_y(self):
        return self.__y
    
    def get_length(self):
        return self.__length
    
    def get_width(self):
        return self.__width
```

7. Add a function to the object classes, including the base, that returns a list of up to 16 pairs of  $x$ and $y$ points on the parameter of the object. 

I used an LLM for this question. I struggled with how to mathematically place points along the perimeter of each shape, especially the circle where trig is involved. I asked the LLM to walk me through the logic step by step and that helped me arrive at the final code. The main thing I took away is that even though all three shapes have a get_perimeter_points() method, the actual math inside each one is completely different depending on the geometry of the shape.

```python
class Shape:
    def __init__(self, x, y):
        self._x = x
        self._y = y
    
    def area(self):
        raise NotImplementedError
    
    def perimeter(self):
        raise NotImplementedError
    
    def get_perimeter_points(self):
        raise NotImplementedError
    
    def get_x(self):
        return self._x
    
    def get_y(self):
        return self._y


class Rectangle(Shape):
    def __init__(self, x, y, length, width):
        Shape.__init__(self, x, y)
        self._length = length
        self._width = width
    
    def area(self):
        return self._length * self._width
    
    def perimeter(self):
        return 2 * (self._length + self._width)
    
    def get_perimeter_points(self):
        points = []
        for i in range(4):
            points.append((self._x, self._y + i * self._width / 3))
        for i in range(4):
            points.append((self._x + i * self._length / 3, self._y + self._width))
        for i in range(4):
            points.append((self._x + self._length, self._y + self._width - i * self._width / 3))
        for i in range(4):
            points.append((self._x + self._length - i * self._length / 3, self._y))
        return points
    
    def get_length(self):
        return self._length
    
    def get_width(self):
        return self._width


class Circle(Shape):
    def __init__(self, x, y, radius):
        Shape.__init__(self, x, y)
        self._radius = radius
    
    def area(self):
        return math.pi * self._radius ** 2
    
    def perimeter(self):
        return 2 * math.pi * self._radius
    
    def get_perimeter_points(self):
        points = []
        for i in range(16):
            angle = 2 * math.pi * i / 16
            px = self._x + self._radius * math.cos(angle)
            py = self._y + self._radius * math.sin(angle)
            points.append((px, py))
        return points
    
    def get_radius(self):
        return self._radius


class Triangle(Shape):
    def __init__(self, x, y, base, height):
        Shape.__init__(self, x, y)
        self._base = base
        self._height = height
    
    def area(self):
        return 0.5 * self._base * self._height
    
    def perimeter(self):
        side = math.sqrt((self._base / 2) ** 2 + self._height ** 2)
        return self._base + 2 * side
    
    def get_perimeter_points(self):
        points = []
        for i in range(6):
            points.append((self._x + i * self._base / 5, self._y))
        for i in range(5):
            t = i / 4
            points.append((self._x + self._base - t * self._base / 2, self._y + t * self._height))
        for i in range(5):
            t = i / 4
            points.append((self._x + self._base / 2 - t * self._base / 2, self._y + self._height - t * self._height))
        return points
    
    def get_base(self):
        return self._base
    
    def get_height(self):
        return self._height
```

```python
r = Rectangle(0, 0, 10, 5)
print("Rectangle points:", r.get_perimeter_points())
print("Number of points:", len(r.get_perimeter_points()))

c = Circle(0, 0, 5)
print("Circle points:", c.get_perimeter_points())
print("Number of points:", len(c.get_perimeter_points()))

t = Triangle(0, 0, 6, 4)
print("Triangle points:", t.get_perimeter_points())
print("Number of points:", len(t.get_perimeter_points()))
```

8. Add a function to the object classes, including the base, that tests if a given set of $x$ and $y$ coordinates are inside of the object. You'll have to think through how to determine if a set of coordinates are inside an object for each object type.

```python
class Shape:
    def __init__(self, x, y):
        self._x = x
        self._y = y
    
    def area(self):
        raise NotImplementedError
    
    def perimeter(self):
        raise NotImplementedError
    
    def get_perimeter_points(self):
        raise NotImplementedError
    
    def is_inside(self, x, y):
        raise NotImplementedError
    
    def get_x(self):
        return self._x
    
    def get_y(self):
        return self._y


class Rectangle(Shape):
    def __init__(self, x, y, length, width):
        Shape.__init__(self, x, y)
        self._length = length
        self._width = width
    
    def area(self):
        return self._length * self._width
    
    def perimeter(self):
        return 2 * (self._length + self._width)
    
    def get_perimeter_points(self):
        points = []
        for i in range(4):
            points.append((self._x, self._y + i * self._width / 3))
        for i in range(4):
            points.append((self._x + i * self._length / 3, self._y + self._width))
        for i in range(4):
            points.append((self._x + self._length, self._y + self._width - i * self._width / 3))
        for i in range(4):
            points.append((self._x + self._length - i * self._length / 3, self._y))
        return points
    
    def is_inside(self, x, y):
        if x >= self._x and x <= self._x + self._length:
            if y >= self._y and y <= self._y + self._width:
                return True
        return False
    
    def get_length(self):
        return self._length
    
    def get_width(self):
        return self._width


class Circle(Shape):
    def __init__(self, x, y, radius):
        Shape.__init__(self, x, y)
        self._radius = radius
    
    def area(self):
        return math.pi * self._radius ** 2
    
    def perimeter(self):
        return 2 * math.pi * self._radius
    
    def get_perimeter_points(self):
        points = []
        for i in range(16):
            angle = 2 * math.pi * i / 16
            px = self._x + self._radius * math.cos(angle)
            py = self._y + self._radius * math.sin(angle)
            points.append((px, py))
        return points
    
    def is_inside(self, x, y):
        distance = math.sqrt((x - self._x) ** 2 + (y - self._y) ** 2)
        return distance <= self._radius
    
    def get_radius(self):
        return self._radius


class Triangle(Shape):
    def __init__(self, x, y, base, height):
        Shape.__init__(self, x, y)
        self._base = base
        self._height = height
    
    def area(self):
        return 0.5 * self._base * self._height
    
    def perimeter(self):
        side = math.sqrt((self._base / 2) ** 2 + self._height ** 2)
        return self._base + 2 * side
    
    def get_perimeter_points(self):
        points = []
        for i in range(6):
            points.append((self._x + i * self._base / 5, self._y))
        for i in range(5):
            t = i / 4
            points.append((self._x + self._base - t * self._base / 2, self._y + t * self._height))
        for i in range(5):
            t = i / 4
            points.append((self._x + self._base / 2 - t * self._base / 2, self._y + self._height - t * self._height))
        return points
    
    def is_inside(self, x, y):
        if y < self._y or y > self._y + self._height:
            return False
        if x < self._x or x > self._x + self._base:
            return False
        
        y_offset = y - self._y
        left_edge = self._x + (self._base / 2) * (1 - y_offset / self._height)
        right_edge = self._x + (self._base / 2) * (1 + y_offset / self._height)
        
        return x >= left_edge and x <= right_edge
    
    def get_base(self):
        return self._base
    
    def get_height(self):
        return self._height
```

```python
r = Rectangle(0, 0, 10, 5)
print("Point (5, 2) in rectangle?", r.is_inside(5, 2))
print("Point (15, 2) in rectangle?", r.is_inside(15, 2))

c = Circle(0, 0, 5)
print("Point (2, 2) in circle?", c.is_inside(2, 2))
print("Point (10, 10) in circle?", c.is_inside(10, 10))

t = Triangle(0, 0, 6, 4)
print("Point (3, 2) in triangle?", t.is_inside(3, 2))
print("Point (10, 10) in triangle?", t.is_inside(10, 10))
```

9. Add a function in the base class of the object classes that returns true/false testing that the object overlaps with another object.

I did have to refer to an LLM for this question. I found it tricky because overlap detection requires two objects to interact with each other, which is a different way of thinking compared to the previous questions that only dealt with a single shape at a time. The LLM helped me figure out the strategy and syntax to check if any of one shape's perimeter points fall inside the other shape, and vice versa, using the get_perimeter_points() and is_inside() methods I had already built.

```python
class Shape:
    def __init__(self, x, y):
        self._x = x
        self._y = y
    
    def area(self):
        raise NotImplementedError
    
    def perimeter(self):
        raise NotImplementedError
    
    def get_perimeter_points(self):
        raise NotImplementedError
    
    def is_inside(self, x, y):
        raise NotImplementedError
    
    def overlaps(self, other):
        points = self.get_perimeter_points()
        for point in points:
            if other.is_inside(point[0], point[1]):
                return True
        
        other_points = other.get_perimeter_points()
        for point in other_points:
            if self.is_inside(point[0], point[1]):
                return True
        
        return False
    
    def get_x(self):
        return self._x
    
    def get_y(self):
        return self._y


class Rectangle(Shape):
    def __init__(self, x, y, length, width):
        Shape.__init__(self, x, y)
        self._length = length
        self._width = width
    
    def area(self):
        return self._length * self._width
    
    def perimeter(self):
        return 2 * (self._length + self._width)
    
    def get_perimeter_points(self):
        points = []
        for i in range(4):
            points.append((self._x, self._y + i * self._width / 3))
        for i in range(4):
            points.append((self._x + i * self._length / 3, self._y + self._width))
        for i in range(4):
            points.append((self._x + self._length, self._y + self._width - i * self._width / 3))
        for i in range(4):
            points.append((self._x + self._length - i * self._length / 3, self._y))
        return points
    
    def is_inside(self, x, y):
        if x >= self._x and x <= self._x + self._length:
            if y >= self._y and y <= self._y + self._width:
                return True
        return False
    
    def get_length(self):
        return self._length
    
    def get_width(self):
        return self._width


class Circle(Shape):
    def __init__(self, x, y, radius):
        Shape.__init__(self, x, y)
        self._radius = radius
    
    def area(self):
        return math.pi * self._radius ** 2
    
    def perimeter(self):
        return 2 * math.pi * self._radius
    
    def get_perimeter_points(self):
        points = []
        for i in range(16):
            angle = 2 * math.pi * i / 16
            px = self._x + self._radius * math.cos(angle)
            py = self._y + self._radius * math.sin(angle)
            points.append((px, py))
        return points
    
    def is_inside(self, x, y):
        distance = math.sqrt((x - self._x) ** 2 + (y - self._y) ** 2)
        return distance <= self._radius
    
    def get_radius(self):
        return self._radius


class Triangle(Shape):
    def __init__(self, x, y, base, height):
        Shape.__init__(self, x, y)
        self._base = base
        self._height = height
    
    def area(self):
        return 0.5 * self._base * self._height
    
    def perimeter(self):
        side = math.sqrt((self._base / 2) ** 2 + self._height ** 2)
        return self._base + 2 * side
    
    def get_perimeter_points(self):
        points = []
        for i in range(6):
            points.append((self._x + i * self._base / 5, self._y))
        for i in range(5):
            t = i / 4
            points.append((self._x + self._base - t * self._base / 2, self._y + t * self._height))
        for i in range(5):
            t = i / 4
            points.append((self._x + self._base / 2 - t * self._base / 2, self._y + self._height - t * self._height))
        return points
    
    def is_inside(self, x, y):
        if y < self._y or y > self._y + self._height:
            return False
        if x < self._x or x > self._x + self._base:
            return False
        
        y_offset = y - self._y
        left_edge = self._x + (self._base / 2) * (1 - y_offset / self._height)
        right_edge = self._x + (self._base / 2) * (1 + y_offset / self._height)
        
        return x >= left_edge and x <= right_edge
    
    def get_base(self):
        return self._base
    
    def get_height(self):
        return self._height
```

10. Copy the `Canvas` class from lecture to in a python file creating a `paint` module. Copy your classes from above into the module and implement paint functions. Implement a `CompoundShape` class. Create a simple drawing demonstrating that all of your classes are working.

```python
class Canvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.data = [[' '] * width for i in range(height)]

    def set_pixel(self, row, col, char='*'):
        self.data[row][col] = char

    def get_pixel(self, row, col):
        return self.data[row][col]

    def clear_canvas(self):
        self.data = [[' '] * self.width for i in range(self.height)]

    def v_line(self, x, y, w, **kargs):
        for i in range(x, x + w):
            self.set_pixel(i, y, **kargs)

    def h_line(self, x, y, h, **kargs):
        for i in range(y, y + h):
            self.set_pixel(x, i, **kargs)

    def line(self, x1, y1, x2, y2, **kargs):
        slope = (y2 - y1) / (x2 - x1)
        for y in range(y1, y2):
            x = int(slope * y)
            self.set_pixel(x, y, **kargs)

    def display(self):
        print("\n".join(["".join(row) for row in self.data]))

class Shape:
    def __init__(self, x, y):
        self._x = x
        self._y = y

    def area(self):
        raise NotImplementedError

    def perimeter(self):
        raise NotImplementedError

    def get_perimeter_points(self):
        raise NotImplementedError

    def is_inside(self, x, y):
        raise NotImplementedError

    def overlaps(self, other):
        for point in self.get_perimeter_points():
            if other.is_inside(point[0], point[1]):
                return True
        for point in other.get_perimeter_points():
            if self.is_inside(point[0], point[1]):
                return True
        return False

    def paint(self, canvas, char='*'):
        pass

    def get_x(self): return self._x
    def get_y(self): return self._y


class Rectangle(Shape):
    def __init__(self, x, y, length, width):
        Shape.__init__(self, x, y)
        self._length = length
        self._width = width

    def area(self):
        return self._length * self._width

    def perimeter(self):
        return 2 * (self._length + self._width)

    def get_perimeter_points(self):
        points = []
        for i in range(4):
            points.append((self._x, self._y + i * self._width / 3))
        for i in range(4):
            points.append((self._x + i * self._length / 3, self._y + self._width))
        for i in range(4):
            points.append((self._x + self._length, self._y + self._width - i * self._width / 3))
        for i in range(4):
            points.append((self._x + self._length - i * self._length / 3, self._y))
        return points

    def is_inside(self, x, y):
        if x >= self._x and x <= self._x + self._length:
            if y >= self._y and y <= self._y + self._width:
                return True
        return False

    def paint(self, canvas, char='*'):
        canvas.v_line(self._x, self._y, self._length, char=char)
        canvas.v_line(self._x, self._y + self._width, self._length, char=char)
        canvas.h_line(self._x, self._y, self._width, char=char)
        canvas.h_line(self._x + self._length, self._y, self._width, char=char)

    def get_length(self): return self._length
    def get_width(self):  return self._width


class Circle(Shape):
    def __init__(self, x, y, radius):
        Shape.__init__(self, x, y)
        self._radius = radius

    def area(self):
        return math.pi * self._radius ** 2

    def perimeter(self):
        return 2 * math.pi * self._radius

    def get_perimeter_points(self):
        points = []
        for i in range(16):
            angle = 2 * math.pi * i / 16
            px = self._x + self._radius * math.cos(angle)
            py = self._y + self._radius * math.sin(angle)
            points.append((px, py))
        return points

    def is_inside(self, x, y):
        distance = math.sqrt((x - self._x) ** 2 + (y - self._y) ** 2)
        return distance <= self._radius

    def paint(self, canvas, char='*'):
        for i in range(360):
            angle = 2 * math.pi * i / 360
            row = int(self._x + self._radius * math.sin(angle))
            col = int(self._y + self._radius * math.cos(angle))
            if 0 <= row < canvas.height and 0 <= col < canvas.width:
                canvas.set_pixel(row, col, char=char)

    def get_radius(self): return self._radius


class Triangle(Shape):
    def __init__(self, x, y, base, height):
        Shape.__init__(self, x, y)
        self._base = base
        self._height = height

    def area(self):
        return 0.5 * self._base * self._height

    def perimeter(self):
        side = math.sqrt((self._base / 2) ** 2 + self._height ** 2)
        return self._base + 2 * side

    def get_perimeter_points(self):
        points = []
        for i in range(6):
            points.append((self._x + i * self._base / 5, self._y))
        for i in range(5):
            t = i / 4
            points.append((self._x + self._base - t * self._base / 2, self._y + t * self._height))
        for i in range(5):
            t = i / 4
            points.append((self._x + self._base / 2 - t * self._base / 2, self._y + self._height - t * self._height))
        return points

    def is_inside(self, x, y):
        if y < self._y or y > self._y + self._height:
            return False
        if x < self._x or x > self._x + self._base:
            return False
        y_offset = y - self._y
        left_edge  = self._x + (self._base / 2) * (1 - y_offset / self._height)
        right_edge = self._x + (self._base / 2) * (1 + y_offset / self._height)
        return x >= left_edge and x <= right_edge

    def paint(self, canvas, char='*'):
        canvas.h_line(self._x + self._height, self._y, self._base, char=char)
        for i in range(self._height):
            offset = int(i * (self._base / 2) / self._height)
            mid = self._y + self._base // 2
            canvas.set_pixel(self._x + i, mid - offset, char=char)
            canvas.set_pixel(self._x + i, mid + offset, char=char)

    def get_base(self):   return self._base
    def get_height(self): return self._height


class CompoundShape(Shape):
    def __init__(self):
        self.shapes = list()

    def add(self, shape):
        self.shapes.append(shape)

    def paint(self, canvas, char='*'):
        for s in self.shapes:
            s.paint(canvas, char=char)
```

```python
canvas = Canvas(60, 25)

r = Rectangle(2, 2, 8, 5)
c = Circle(5, 35, 8)
t = Triangle(14, 5, 20, 10)

comp = CompoundShape()
comp.add(r)
comp.add(c)
comp.add(t)

comp.paint(canvas, char='*')

canvas.display()
```

11. Create a `RasterDrawing` class. Demonstrate that you can create a drawing made of several shapes, paint the drawing, modify the drawing, and paint it again. 

```python
class RasterDrawing:
    def __init__(self, width, height):
        self.canvas = Canvas(width, height)
        self.shapes = list()

    def add_shape(self, shape):
        self.shapes.append(shape)

    def remove_shape(self, shape):
        if shape in self.shapes:
            self.shapes.remove(shape)

    def paint(self, char='*'):
        self.canvas.clear_canvas()
        for s in self.shapes:
            s.paint(self.canvas, char=char)
        self.canvas.display()
```

```python
rd = RasterDrawing(60, 25)

r = Rectangle(2, 2, 8, 5)
rd.add_shape(r)
rd.add_shape(Circle(5, 35, 8))
rd.add_shape(Triangle(14, 5, 20, 10))

print("1st paint")
rd.paint(char='*')

r._length = 15
r._width = 10

print("2nd paint")
rd.paint(char='*')
```

12. Implement the ability to load/save raster drawings and demonstate that your method works. One way to implement this ability:

   * Overload `__repr__` functions of all objects to return strings of the python code that would construct the object.
   
   * In the save method of raster drawing class, store the representations into the file.
   * Write a loader function that reads the file and uses `eval` to instantiate the object.

For example:

I had to refer to an LLM for this question. I found it difficult in the very beginning because I had never seen the idea of turning objects into strings before. The LLM walked me through how __repr__ can be overloaded to return a string of the exact python code needed to reconstruct the object, so that eval() can just run that string and hand the object back. I also ran into a bug where my __repr__ wasn't being picked up because I defined it in a separate cell rather than inside the original class, and the LLM explained how to fix that. It went over the syntax and everythings general purpose as well. 

```python
class foo:
    def __init__(self,a,b=None):
        self.a=a
        self.b=b
        
    def __repr__(self):
        return "foo("+repr(self.a)+","+repr(self.b)+")"
    
    def save(self,filename):
        f=open(filename,"w")
        f.write(self.__repr__())
        f.close()
        
   
def foo_loader(filename):
    f=open(filename,"r")
    tmp=eval(f.read())
    f.close()
    return tmp
```

```python
class Rectangle(Shape):
    def __repr__(self):
        return "Rectangle(" + repr(self._x) + "," + repr(self._y) + "," + repr(self._length) + "," + repr(self._width) + ")"

class Circle(Shape):
    def __repr__(self):
        return "Circle(" + repr(self._x) + "," + repr(self._y) + "," + repr(self._radius) + ")"

class Triangle(Shape):
    def __repr__(self):
        return "Triangle(" + repr(self._x) + "," + repr(self._y) + "," + repr(self._base) + "," + repr(self._height) + ")"
```

```python
def rect_repr(self):
    return "Rectangle(" + repr(self._x) + "," + repr(self._y) + "," + repr(self._length) + "," + repr(self._width) + ")"

def circle_repr(self):
    return "Circle(" + repr(self._x) + "," + repr(self._y) + "," + repr(self._radius) + ")"

def triangle_repr(self):
    return "Triangle(" + repr(self._x) + "," + repr(self._y) + "," + repr(self._base) + "," + repr(self._height) + ")"

Rectangle.__repr__ = rect_repr
Circle.__repr__ = circle_repr
Triangle.__repr__ = triangle_repr

class RasterDrawing:
    def __init__(self, width, height):
        self.canvas = Canvas(width, height)
        self.shapes = list()

    def add_shape(self, shape):
        self.shapes.append(shape)

    def remove_shape(self, shape):
        if shape in self.shapes:
            self.shapes.remove(shape)

    def paint(self, char='*'):
        self.canvas.clear_canvas()
        for s in self.shapes:
            s.paint(self.canvas, char=char)
        self.canvas.display()

    def save(self, filename):
        f = open(filename, "w")
        for s in self.shapes:
            f.write(repr(s) + "\n")
        f.close()


def load_drawing(filename, width, height):
    rd = RasterDrawing(width, height)
    f = open(filename, "r")
    for line in f:
        shape = eval(line)
        rd.add_shape(shape)
    f.close()
    return rd
```

```python
rd = RasterDrawing(60, 25)
rd.add_shape(Rectangle(2, 2, 8, 5))
rd.add_shape(Circle(5, 35, 8))
rd.add_shape(Triangle(14, 5, 20, 10))

print("original")
rd.paint(char='*')

rd.save("my_drawing.txt")

!cat my_drawing.txt

rd2 = load_drawing("my_drawing.txt", 60, 25)
print("loaded")
rd2.paint(char='*')
```
