1. Create a `matrix` class with the following properties:
    * It can be initialized in 2 ways:
        1. with arguments `n` and `m`, the size of the matrix. A newly instanciated matrix will contain all zeros.
        2. with a list of lists of values. Note that since we are using lists of lists to implement matrices, it is possible that not all rows have the same number of columns. Test explicitly that the matrix is properly specified.
    * Matrix instances `M` can be indexed with `M[i][j]` and `M[i,j]`.
    * Matrix assignment works in 2 ways:
        1. If `M_1` and `M_2` are `matrix` instances `M_1=M_2` sets the values of `M_1` to those of `M_2`, if they are the same size. Error otherwise.
        2. In example above `M_2` can be a list of lists of correct size.


```python
def zero_matrix(m, n):
    return [[0 for _ in range(n)] for _ in range(m)]

def is_matrix(M):
    if isinstance(M, list):
        row_length = len(M[0])
        for row in M:
            if not row_length == len(row):
                return False
    else:
        return False
    return True

def matrix_shape(M):
    if is_matrix(M):
        m = len(M)
        n = len(M[0])
        return m, n
    else:
        return 0, 0


class matrix:

    def __init__(self, n_or_list=None, m=None):
        if isinstance(n_or_list, list):
            if not is_matrix(n_or_list):
                raise ValueError("Error")
            self._data = [row[:] for row in n_or_list]
            self._m, self._n = matrix_shape(self._data)
        elif isinstance(n_or_list, int) and isinstance(m, int):
            self._m = n_or_list
            self._n = m
            self._data = zero_matrix(self._m, self._n)
        else:
            raise ValueError("Error")

    def shape(self):
        return self._m, self._n

    def __getitem__(self, key):
        if isinstance(key, int):
            return self._data[key]
        elif isinstance(key, tuple):
            i, j = key
            return self._data[i][j]
        else:
            raise TypeError("Error")

    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self._data[i][j] = value
        elif isinstance(key, int):
            if isinstance(value, list) and len(value) == self._n:
                self._data[key] = value[:]
            else:
                raise ValueError("Error".format(self._n))
        else:
            raise TypeError("Error")

    def set(self, other):
        if isinstance(other, matrix):
            if other.shape() != self.shape():
                raise ValueError("Error".format(self.shape(), other.shape()))
            self._data = [row[:] for row in other._data]
        elif isinstance(other, list):
            if not is_matrix(other):
                raise ValueError("Error")
            if matrix_shape(other) != self.shape():
                raise ValueError("Error".format(self.shape(), matrix_shape(other)))
            self._data = [row[:] for row in other]
        else:
            raise TypeError("Error")

    def __repr__(self):
        rows = ["  " + str(row) for row in self._data]
        return "matrix([\n" + ",\n".join(rows) + "\n])"
```

```python
M1 = matrix(3, 3)
print(M1)

M2 = matrix([[1, 2, 3], [4, 5, 6], [7, 8, 9]])
print(M2)

try:
    matrix([[1, 2, 3], [4, 5]])
except ValueError as e:
    print(e)

print(M2[1][2])
print(M2[1, 2])

M2[1, 1] = 99
M2[0][0] = 42
print(M2)

M3 = matrix(3, 3)
M3.set(M2)
print(M3)

M4 = matrix(3, 3)
M4.set([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print(M4)

try:
    matrix(2, 2).set(M2)
except ValueError as e:
    print(e)
```

2. Add the following methods:
    * `shape()`: returns a tuple `(n,m)` of the shape of the matrix.
    * `transpose()`: returns a new matrix instance which is the transpose of the matrix.
    * `row(n)` and `column(n)`: that return the nth row or column of the matrix M as a new appropriately shaped matrix object.
    * `to_list()`: which returns the matrix as a list of lists.
    *  `block(n_0,n_1,m_0,m_1)` that returns a smaller matrix located at the n_0 to n_1 columns and m_0 to m_1 rows. 
    * Modify `__getitem__` implemented above to support slicing.
        

```python
def zero_matrix(m, n):
    return [[0 for _ in range(n)] for _ in range(m)]

def is_matrix(M):
    if isinstance(M, list):
        row_length = len(M[0])
        for row in M:
            if not row_length == len(row):
                return False
    else:
        return False
    return True

def matrix_shape(M):
    if is_matrix(M):
        m = len(M)
        n = len(M[0])
        return m, n
    else:
        return 0, 0


class matrix:

    def __init__(self, n_or_list=None, m=None):
        if isinstance(n_or_list, list):
            if not is_matrix(n_or_list):
                raise ValueError("Error")
            self._data = [row[:] for row in n_or_list]
            self._m, self._n = matrix_shape(self._data)
        elif isinstance(n_or_list, int) and isinstance(m, int):
            self._m = n_or_list
            self._n = m
            self._data = zero_matrix(self._m, self._n)
        else:
            raise ValueError("Error")

    def shape(self):
        return self._m, self._n

    def transpose(self):
        transposed = zero_matrix(self._n, self._m)
        for i in range(self._m):
            for j in range(self._n):
                transposed[j][i] = self._data[i][j]
        return matrix(transposed)

    def row(self, n):
        return matrix([self._data[n]])

    def column(self, n):
        col = [[self._data[i][n]] for i in range(self._m)]
        return matrix(col)

    def to_list(self):
        return [row[:] for row in self._data]

    def block(self, n_0, n_1, m_0, m_1):
        sub = [self._data[i][m_0:m_1] for i in range(n_0, n_1)]
        return matrix(sub)

    def __getitem__(self, key):
        if isinstance(key, int):
            return self._data[key]
        elif isinstance(key, tuple):
            i, j = key
            return self._data[i][j]
        elif isinstance(key, slice):
            start = key.start or 0
            stop = key.stop or self._m
            step = key.step or 1
            return matrix([self._data[i] for i in range(start, stop, step)])
        else:
            raise TypeError("Error")

    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self._data[i][j] = value

    def set(self, other):
        if isinstance(other, matrix):
            if other.shape() != self.shape():
                raise ValueError("Error".format(self.shape(), other.shape()))
            self._data = [row[:] for row in other._data]
        elif isinstance(other, list):
            if not is_matrix(other):
                raise ValueError("Error")
            if matrix_shape(other) != self.shape():
                raise ValueError("Error".format(self.shape(), matrix_shape(other)))
            self._data = [row[:] for row in other]
        else:
            raise TypeError("Error")

    def __repr__(self):
        rows = ["  " + str(row) for row in self._data]
        return "matrix([\n" + ",\n".join(rows) + "\n])"
```

```python
M = matrix([[1, 2, 3], [4, 5, 6], [7, 8, 9]])

print(M.shape())
print(M.transpose())
print(M.row(1))
print(M.column(2))
print(M.to_list())
print(M.block(0, 2, 1, 3))
print(M[1:3])
```

3. Write functions that create special matrices (note these are standalone functions, not member functions of your `matrix` class):
    * `constant(n,m,c)`: returns a `n` by `m` matrix filled with floats of value `c`.
    * `zeros(n,m)` and `ones(n,m)`: return `n` by `m` matrices filled with floats of value `0` and `1`, respectively.
    * `eye(n)`: returns the n by n identity matrix.

```python
def constant(n, m, c):
    return matrix([[float(c) for _ in range(m)] for _ in range(n)])

def zeros(n, m):
    return constant(n, m, 0)

def ones(n, m):
    return constant(n, m, 1)

def eye(n):
    result = zeros(n, n)
    for i in range(n):
        result[i, i] = 1.0
    return result
```

```python
print(constant(2, 3, 5))
print(zeros(2, 3))
print(ones(2, 3))
print(eye(3))
```

4. Add the following member functions to your class. Make sure to appropriately test the dimensions of the matrices to make sure the operations are correct.
    * `M.scalarmul(c)`: a matrix that is scalar product $cM$, where every element of $M$ is multiplied by $c$.
    * `M.add(N)`: adds two matrices $M$ and $N$. Don’t forget to test that the sizes of the matrices are compatible for this and all other operations.
    * `M.sub(N)`: subtracts two matrices $M$ and $N$.
    * `M.mat_mult(N)`: returns a matrix that is the matrix product of two matrices $M$ and $N$.
    * `M.element_mult(N)`: returns a matrix that is the element-wise product of two matrices $M$ and $N$.
    * `M.equals(N)`: returns true/false if $M==N$.

```python

class matrix:

    def __init__(self, n_or_list=None, m=None):
        if isinstance(n_or_list, list):
            if not is_matrix(n_or_list):
                raise ValueError("Error")
            self._data = [row[:] for row in n_or_list]
            self._m, self._n = matrix_shape(self._data)
        elif isinstance(n_or_list, int) and isinstance(m, int):
            self._m = n_or_list
            self._n = m
            self._data = zero_matrix(self._m, self._n)
        else:
            raise ValueError("Error")

    def shape(self):
        return self._m, self._n

    def transpose(self):
        transposed = zero_matrix(self._n, self._m)
        for i in range(self._m):
            for j in range(self._n):
                transposed[j][i] = self._data[i][j]
        return matrix(transposed)

    def row(self, n):
        return matrix([self._data[n]])

    def column(self, n):
        col = [[self._data[i][n]] for i in range(self._m)]
        return matrix(col)

    def to_list(self):
        return [row[:] for row in self._data]

    def block(self, n_0, n_1, m_0, m_1):
        sub = [self._data[i][m_0:m_1] for i in range(n_0, n_1)]
        return matrix(sub)

    def __getitem__(self, key):
        if isinstance(key, int):
            return self._data[key]
        elif isinstance(key, tuple):
            i, j = key
            return self._data[i][j]
        elif isinstance(key, slice):
            start = key.start or 0
            stop = key.stop or self._m
            step = key.step or 1
            return matrix([self._data[i] for i in range(start, stop, step)])
        else:
            raise TypeError("Error")

    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self._data[i][j] = value

    def set(self, other):
        if isinstance(other, matrix):
            if other.shape() != self.shape():
                raise ValueError("Error".format(self.shape(), other.shape()))
            self._data = [row[:] for row in other._data]
        elif isinstance(other, list):
            if not is_matrix(other):
                raise ValueError("Error")
            if matrix_shape(other) != self.shape():
                raise ValueError("Error".format(self.shape(), matrix_shape(other)))
            self._data = [row[:] for row in other]
        else:
            raise TypeError("Error")

    def scalarmul(self, c):
        result = zero_matrix(self._m, self._n)
        for i in range(self._m):
            for j in range(self._n):
                result[i][j] = self._data[i][j] * c
        return matrix(result)

    def add(self, N):
        if self.shape() != N.shape():
            raise ValueError("Error".format(self.shape(), N.shape()))
        result = zero_matrix(self._m, self._n)
        for i in range(self._m):
            for j in range(self._n):
                result[i][j] = self._data[i][j] + N[i][j]
        return matrix(result)

    def sub(self, N):
        if self.shape() != N.shape():
            raise ValueError("Error".format(self.shape(), N.shape()))
        result = zero_matrix(self._m, self._n)
        for i in range(self._m):
            for j in range(self._n):
                result[i][j] = self._data[i][j] - N[i][j]
        return matrix(result)

    def mat_mult(self, N):
        if self._n != N.shape()[0]:
            raise ValueError("Error".format(self.shape(), N.shape()))
        result = zero_matrix(self._m, N.shape()[1])
        for i in range(self._m):
            for j in range(N.shape()[1]):
                for k in range(self._n):
                    result[i][j] += self._data[i][k] * N[k][j]
        return matrix(result)

    def element_mult(self, N):
        if self.shape() != N.shape():
            raise ValueError("Error".format(self.shape(), N.shape()))
        result = zero_matrix(self._m, self._n)
        for i in range(self._m):
            for j in range(self._n):
                result[i][j] = self._data[i][j] * N[i][j]
        return matrix(result)

    def equals(self, N):
        if self.shape() != N.shape():
            return False
        for i in range(self._m):
            for j in range(self._n):
                if self._data[i][j] != N[i][j]:
                    return False
        return True

    def __repr__(self):
        rows = ["  " + str(row) for row in self._data]
        return "matrix([\n" + ",\n".join(rows) + "\n])"
```

```python
A = matrix([[1, 2], [3, 4]])
B = matrix([[5, 6], [7, 8]])

print(A.scalarmul(2))
print(A.add(B))
print(A.sub(B))
print(A.mat_mult(B))
print(A.element_mult(B))
print(A.equals(A))
print(A.equals(B))

try:
    A.add(matrix(3, 3))
except ValueError as e:
    print(e)
```

5. Overload python operators to appropriately use your functions in 4 and allow expressions like:
    * 2*M
    * M*2
    * M+N
    * M-N
    * M*N
    * M==N
    * M=N


```python
def zero_matrix(m, n):
    return [[0 for _ in range(n)] for _ in range(m)]

def is_matrix(M):
    if isinstance(M, list):
        row_length = len(M[0])
        for row in M:
            if not row_length == len(row):
                return False
    else:
        return False
    return True

def matrix_shape(M):
    if is_matrix(M):
        m = len(M)
        n = len(M[0])
        return m, n
    else:
        return 0, 0


class matrix:

    def __init__(self, n_or_list=None, m=None):
        if isinstance(n_or_list, list):
            if not is_matrix(n_or_list):
                raise ValueError("Error")
            self._data = [row[:] for row in n_or_list]
            self._m, self._n = matrix_shape(self._data)
        elif isinstance(n_or_list, int) and isinstance(m, int):
            self._m = n_or_list
            self._n = m
            self._data = zero_matrix(self._m, self._n)
        else:
            raise ValueError("Error")

    def shape(self):
        return self._m, self._n

    def transpose(self):
        transposed = zero_matrix(self._n, self._m)
        for i in range(self._m):
            for j in range(self._n):
                transposed[j][i] = self._data[i][j]
        return matrix(transposed)

    def row(self, n):
        return matrix([self._data[n]])

    def column(self, n):
        col = [[self._data[i][n]] for i in range(self._m)]
        return matrix(col)

    def to_list(self):
        return [row[:] for row in self._data]

    def block(self, n_0, n_1, m_0, m_1):
        sub = [self._data[i][m_0:m_1] for i in range(n_0, n_1)]
        return matrix(sub)

    def __getitem__(self, key):
        if isinstance(key, int):
            return self._data[key]
        elif isinstance(key, tuple):
            i, j = key
            return self._data[i][j]
        elif isinstance(key, slice):
            start = key.start or 0
            stop = key.stop or self._m
            step = key.step or 1
            return matrix([self._data[i] for i in range(start, stop, step)])
        else:
            raise TypeError("Error")

    def __setitem__(self, key, value):
        if isinstance(key, tuple):
            i, j = key
            self._data[i][j] = value

    def set(self, other):
        if isinstance(other, matrix):
            if other.shape() != self.shape():
                raise ValueError("Error".format(self.shape(), other.shape()))
            self._data = [row[:] for row in other._data]
        elif isinstance(other, list):
            if not is_matrix(other):
                raise ValueError("Error")
            if matrix_shape(other) != self.shape():
                raise ValueError("Error".format(self.shape(), matrix_shape(other)))
            self._data = [row[:] for row in other]
        else:
            raise TypeError("Error")

    def scalarmul(self, c):
        result = zero_matrix(self._m, self._n)
        for i in range(self._m):
            for j in range(self._n):
                result[i][j] = self._data[i][j] * c
        return matrix(result)

    def add(self, N):
        if self.shape() != N.shape():
            raise ValueError("Error".format(self.shape(), N.shape()))
        result = zero_matrix(self._m, self._n)
        for i in range(self._m):
            for j in range(self._n):
                result[i][j] = self._data[i][j] + N[i][j]
        return matrix(result)

    def sub(self, N):
        if self.shape() != N.shape():
            raise ValueError("Error".format(self.shape(), N.shape()))
        result = zero_matrix(self._m, self._n)
        for i in range(self._m):
            for j in range(self._n):
                result[i][j] = self._data[i][j] - N[i][j]
        return matrix(result)

    def mat_mult(self, N):
        if self._n != N.shape()[0]:
            raise ValueError("Error".format(self.shape(), N.shape()))
        result = zero_matrix(self._m, N.shape()[1])
        for i in range(self._m):
            for j in range(N.shape()[1]):
                for k in range(self._n):
                    result[i][j] += self._data[i][k] * N[k][j]
        return matrix(result)

    def element_mult(self, N):
        if self.shape() != N.shape():
            raise ValueError("Error".format(self.shape(), N.shape()))
        result = zero_matrix(self._m, self._n)
        for i in range(self._m):
            for j in range(self._n):
                result[i][j] = self._data[i][j] * N[i][j]
        return matrix(result)

    def equals(self, N):
        if self.shape() != N.shape():
            return False
        for i in range(self._m):
            for j in range(self._n):
                if self._data[i][j] != N[i][j]:
                    return False
        return True

    def __mul__(self, other):
        if isinstance(other, matrix):
            return self.mat_mult(other)
        else:
            return self.scalarmul(other)

    def __rmul__(self, other):
        return self.scalarmul(other)

    def __add__(self, other):
        return self.add(other)

    def __sub__(self, other):
        return self.sub(other)

    def __eq__(self, other):
        return self.equals(other)

    def __repr__(self):
        rows = ["  " + str(row) for row in self._data]
        return "matrix([\n" + ",\n".join(rows) + "\n])"
```

```python
A = matrix([[1, 2], [3, 4]])
B = matrix([[5, 6], [7, 8]])

print(2 * A)
print(A * 2)
print(A + B)
print(A - B)
print(A * B)
print(A == B)
print(A == A)
```
