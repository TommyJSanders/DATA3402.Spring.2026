```python
# Part a: Complete low-level features pair plot (1 cell)
def compare_distributions(df, feature, alpha=0.5, density=1, bins=50):
    """Diagonal histograms: signal vs background"""
    plt.hist(df[df['signal']==1][feature], bins=bins, alpha=alpha, label='Signal', density=density)
    plt.hist(df[df['signal']==0][feature], bins=bins, alpha=alpha, label='Background', density=density)
    plt.legend()

def compare_scatter(df, x_var, y_var, make_legend=True):
    """Off-diagonal scatter: signal (blue) vs background (red)"""
    plt.scatter(df[df['signal']==1][x_var], df[df['signal']==1][y_var], alpha=0.3, s=1, c='blue', label='Signal')
    plt.scatter(df[df['signal']==0][x_var], df[df['signal']==0][y_var], alpha=0.3, s=1, c='red', label='Background')
    if make_legend: plt.legend()

# Low-level raw features pair plot like Lecture 13
columns = df.columns[1:6]  # l_1_pT through l_2_eta (5 raw kinematics)
n_columns = len(columns)
plt.figure(figsize=(15,15))

plot_i = 0
for i, x_var_name in enumerate(columns):
    for j, y_var_name in enumerate(columns):
        plot_i += 1
        plt.subplot(n_columns, n_columns, plot_i)
        make_legend = plot_i == 1  # Legend only on first subplot
        if i == j:
            compare_distributions(df, x_var_name)
        else:
            compare_scatter(df, x_var_name, y_var_name, make_legend=make_legend)

plt.tight_layout()
plt.show()
```

## 5.1

Most variables favor $x > x_c$ since signal tends to peak at higher values than background. The exceptions are
dPhi_r_b, which favors $x < x_c$, and
l_1_eta/l_2_eta, which favor $|x - \mu| < x_c$ since signal is more tightly concentrated around 0. The phi variables (
l_1_phi, l_2_phi, MET_phi) offer little to no separation power under any strategy.

## 5.2

```python
# Compute cumulative rate (TPR or FPR) for a given observable
def compute_rate(d, bins=100):
    hist, bins_ = np.histogram(d, bins=bins, density=True)
    # Reverse cumulative sum to get fraction of events passing x > x_c
    R = np.cumsum(hist[::-1])[::-1] * (bins_[1] - bins_[0])
    return R, bins_

# Loop over all observables and plot TPR and FPR for each
for obs in VarNames[1:]:
    TPR, bins_ = compute_rate(df_sig[obs])
    FPR, bins_ = compute_rate(df_bkg[obs], bins=bins_)
    
    plt.figure(figsize=(8,4))
    plt.plot(bins_[:-1], TPR, label="TPR (Signal Efficiency)")
    plt.plot(bins_[:-1], FPR, label="FPR (Background Efficiency)")
    plt.xlabel(obs)
    plt.ylabel("Rate")
    plt.title(obs)
    plt.legend()
    plt.show()
```

## 5.3 

```python
# Four scenarios with different expected signal and background counts
scenarios = { "1": (10, 100), 
              "2": (100, 1000),
              "3": (1000, 10000),
              "4": (10000, 100000)
            }

def compare_significance(df_sig, df_bkg, obs_name, scenarios, bins=100, log=False):
    # Compute TPR and FPR for the given observable
    TPR, bins_ = compute_rate(df_sig[obs_name], bins=bins)
    FPR, bins_ = compute_rate(df_bkg[obs_name], bins=bins_)
    
    for name, (n_sig_expected, n_bkg_expected) in scenarios.items():
        # Scale by expected counts to get N'_S and N'_B
        n_sig_prime = n_sig_expected * TPR
        n_bkg_prime = n_bkg_expected * FPR
        
        # Compute significance and plot as function of x_c
        sig = n_sig_prime / np.sqrt(n_sig_prime + n_bkg_prime)
        plt.step(bins_[:-1], sig, label=name + " " + str((n_sig_expected, n_bkg_expected)))
    
    if log:
        plt.yscale("log")
    plt.xlabel(obs_name)
    plt.ylabel("Significance")
    plt.title(obs_name)
    plt.legend()
    plt.show()

# Plot significance curves for all observables
for obs in VarNames[1:]:
    compare_significance(df_sig, df_bkg, obs, scenarios)
```

## 6.1

Based on the distributions identified in Exercise 4.1c, the three observables chosen for selection are M_TR_2, M_R, and MT2, as these showed the clearest separation between signal and background. The optimal cut values xcx_c
xc based on the significance peaks from 5.3 are approximately xc≈1.35 for M_TR_2, and xc≈1.0  for both M_R and MT2.

## 6.2 

```python
# Four scenarios with different expected signal and background counts
scenarios = {"1": (10, 100), 
             "2": (100, 1000),
             "3": (1000, 10000),
             "4": (10000, 100000)}

# Chosen observables and their cut values from 5.3
cuts = {"M_TR_2": 1.35, "M_R": 1.0, "MT2": 1.0}

for scenario_name, (N_S, N_B) in scenarios.items():
    print(f"Scenario {scenario_name}: N_S={N_S}, N_B={N_B}")
    
    # Start with full dataset and apply cuts successively
    df_sig_cut = df_sig.copy()
    df_bkg_cut = df_bkg.copy()
    
    table = []
    for obs, xc in cuts.items():
        # Apply cut on current observable
        df_sig_cut = df_sig_cut[df_sig_cut[obs] > xc]
        df_bkg_cut = df_bkg_cut[df_bkg_cut[obs] > xc]
        
        # Compute efficiencies, scaled counts, and significance
        eps_S = df_sig_cut.shape[0] / df_sig.shape[0]
        eps_B = df_bkg_cut.shape[0] / df_bkg.shape[0]
        N_S_prime = eps_S * N_S
        N_B_prime = eps_B * N_B
        sigma = N_S_prime / np.sqrt(N_S_prime + N_B_prime)
        
        table.append([obs, round(eps_S,4), round(eps_B,4), round(N_S_prime,4), round(N_B_prime,4), round(sigma,4)])
    
    # Display cut-flow table for this scenario
    display(HTML(tabulate.tabulate(table, tablefmt='html', headers=["Observable", "eps_S", "eps_B", "N_S'", "N_B'", "sigma"])))
```

## 6.3 

Looking at the correlation matrix from Exercise 4.2, M_TR_2 and M_R show the highest correlation among the top separating variables. This means that cutting on one will affect the optimal xc for the other, making the independent significance estimates from 5.3 non-optimal.

```python
# Apply cut on M_TR_2 first, then optimize M_R
df_sig_1 = df_sig[df_sig["M_TR_2"] > 1.35]
df_bkg_1 = df_bkg[df_bkg["M_TR_2"] > 1.35]

eff_s = df_sig_1.shape[0] / df_sig.shape[0]
eff_b = df_bkg_1.shape[0] / df_bkg.shape[0]

scenarios_1 = dict(zip(scenarios.keys(), map(lambda x: (eff_s*x[0], eff_b*x[1]), scenarios.values())))

compare_significance(df_sig_1, df_bkg_1, "M_R", scenarios_1)

# Reverse order: cut on M_R first, then optimize M_TR_2
df_sig_2 = df_sig[df_sig["M_R"] > 1.0]
df_bkg_2 = df_bkg[df_bkg["M_R"] > 1.0]

eff_s = df_sig_2.shape[0] / df_sig.shape[0]
eff_b = df_bkg_2.shape[0] / df_bkg.shape[0]

scenarios_2 = dict(zip(scenarios.keys(), map(lambda x: (eff_s*x[0], eff_b*x[1]), scenarios.values())))

compare_significance(df_sig_2, df_bkg_2, "M_TR_2", scenarios_2)
```

Changing the order of the selections produces different significance curves for the second variable, with the optimal xc shifting depending on which cut was applied first. This confirms that the correlation between
M_TR_2 and M_R means the two variables are not independent, so the order of cuts matters when trying to maximize significance.

```python
# AUC function from Lecture 17
def AUC(TPR, FPR):
    n = FPR.shape[0]
    return np.sum((FPR[0:n-1] - FPR[1:]) * (TPR[0:n-1] + TPR[1:]) / 2)

# Top 3 observables from Exercise 4.1c
top_obs = ["M_TR_2", "M_R", "MT2"]

for obs in top_obs:
    # Compute TPR and FPR for each observable
    TPR, bins_ = compute_rate(df_sig[obs])
    FPR, bins_ = compute_rate(df_bkg[obs], bins=bins_)
    
    # Compute area under the ROC curve
    auc = AUC(TPR, FPR)
    
    # Plot ROC curve with AUC in the legend
    plt.plot(FPR, TPR, label=f"{obs} (AUC={round(auc, 3)})")

# Add diagonal reference line for random classifier
plt.plot([0,1], [0,1], 'k--', label="Random")
plt.xlabel("FPR")
plt.ylabel("TPR")
plt.title("ROC Curves")
plt.legend()
plt.show()
```

```python
def plot_roc_curves(df_sig, df_bkg, obs_list):
    # Loop over each observable and plot its ROC curve
    for obs in obs_list:
        # Compute TPR and FPR
        TPR, bins_ = compute_rate(df_sig[obs])
        FPR, bins_ = compute_rate(df_bkg[obs], bins=bins_)
        
        # Compute AUC and plot with label
        auc = AUC(TPR, FPR)
        plt.plot(FPR, TPR, label=f"{obs} (AUC={round(auc, 3)})")
    
    # Add diagonal reference line for random classifier
    plt.plot([0,1], [0,1], 'k--', label="Random")
    plt.xlabel("FPR")
    plt.ylabel("TPR")
    plt.title("ROC Curves")
    plt.legend()
    plt.show()

# Test with top 3 observables
plot_roc_curves(df_sig, df_bkg, ["M_TR_2", "M_R", "MT2"])
```

```python
df_sig_0 = df_sig.drop("signal", axis=1)
df_bkg_0 = df_bkg.drop("signal", axis=1)

# Compute mean vector for signal and background
m_s = np.mean(df_sig_0, axis=0)
m_b = np.mean(df_bkg_0, axis=0)

# Compute delta as a column vector for outer product
delta = np.matrix(m_s - m_b).transpose()

# Compute between class covariance matrix S_B = delta * delta^T
S_B = delta * delta.transpose()

# Compute within-class covariance matrix for signal
delta_s = np.matrix(df_sig_0 - m_s).transpose()
S_W_s = delta_s * delta_s.transpose()

# Compute within-class covariance matrix for background
delta_b = np.matrix(df_bkg_0 - m_b).transpose()
S_W_b = delta_b * delta_b.transpose()

S_W = S_W_s + S_W_b

print("S_B shape:", S_B.shape)
print("S_W shape:", S_W.shape)
```

```python
S_W_inv = np.linalg.inv(S_W)

# Compute linear coefficients w = S_W^-1 * (m_b - m_s)
w = S_W_inv * np.matrix(m_b - m_s).transpose()

# Normalize weights so they sum to 1
w_1 = w / sum(w)

# Apply weights to get discriminant output for signal and background
output_s = np.matrix(df_sig_0) * w_1
output_b = np.matrix(df_bkg_0) * w_1

# Plot histograms for signal and background
h_s, bins, _ = plt.hist(output_s, label="signal", alpha=0.5, bins=100)
h_b, bins, _ = plt.hist(output_b, bins=bins, alpha=0.5, label="background")
plt.legend()
plt.show()
```
